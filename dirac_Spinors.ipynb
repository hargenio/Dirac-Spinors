```
(* ========================================================================= *)
(* APÉNDICE: Validación Computacional de Espinores con FeynCalc              *)
(* Tesina: Relaciones entre espinores u y v en la ecuación de Dirac          *)
(* ========================================================================= *)

(* 1. Inicialización del paquete *)
If[$FeynCalcLoaded === False, << FeynCalc`];

(* Definición de variables y dimensiones *)
SetMandelstam[s, t, u, p1, p2, k1, k2, m, 0];
Print["--- INICIO DE VALIDACIÓN DE ESPINORES ---"];

(* ========================================================================= *)
(* 2. Verificación de la Ecuación de Dirac para partículas (u) y antipartículas (v) *)
(* ========================================================================= *)

Print["\n[1] Ecuaciones de movimiento fundamentales:"];

(* Espinor u: p_slash . u(p) = m u(p) *)
eqU = DiracSimplify[GS[p] . SpinorU[p, m]];
Print["Dirac con u(p): ", eqU];

(* Espinor v: p_slash . v(p) = -m v(p) *)
eqV = DiracSimplify[GS[p] . SpinorV[p, m]];
Print["Dirac con v(p): ", eqV];

```

```


(* ========================================================================= *)
(* 3. Relaciones de Normalización (Normas covariantes)                        *)
(* ========================================================================= *)

Print["\n[2] Normas de los espinores de Dirac:"];

(* Norma del espinor u: u_barra . u = 2m *)
normU = DiracSimplify[SpinorUBar[p, m] . SpinorU[p, m]];
Print["Norma ubar . u = ", normU];

(* Norma del espinor v: v_barra . v = -2m *)
normV = DiracSimplify[SpinorVBar[p, m] . SpinorV[p, m]];
Print["Norma vbar . v = ", normV];

```

```

(* ========================================================================= *)
(* 4. Ortogonalidad Cruzada entre estados de energía positiva y negativa     *)
(* ========================================================================= *)

Print["\n[3] Ortogonalidad cruzada:"];

(* Producto u_barra . v evaluado al mismo momento = 0 *)
ortogUV = DiracSimplify[SpinorUBar[p, m] . SpinorV[p, m]];
Print["Ortogonalidad ubar . v = ", ortogUV];

(* Producto v_barra . u evaluado al mismo momento = 0 *)
ortogVU = DiracSimplify[SpinorVBar[p, m] . SpinorU[p, m]];
Print["Ortogonalidad vbar . u = ", ortogVU];

```

```

(* ========================================================================= *)
(* 5. Relaciones de Completitud (Sumas sobre espines)                        *)
(* ========================================================================= *)

Print["\n[4] Relaciones de completitud (Sumas sobre espines):"];

(* Suma sobre espines para u expandida *)
sumU = SpinorSum[p, m] /. SpinorSum[q_, mass_] :> GS[q] + mass;
Print["SpinorSum para u expandida: ", sumU];

(* Suma sobre espines para v expandida *)
sumV = SpinorSum[p, -m] /. SpinorSum[q_, mass_] :> GS[q] + mass;
Print["SpinorSum para v expandida: ", sumV];

Print["\n--- FIN DE LA VALIDACIÓN COMPUTACIONAL ---"];
```
